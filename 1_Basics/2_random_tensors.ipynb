# Random Tensors

`Tensors` are basic building blocks in `tensorflow`. But sometimes we need to create random tensors. In this notebook, we will see how to create random tensors in `tensorflow`.

So, how do we create random tensors in `tensorflow`?

```python
import tensorflow as tf
```

```python
tf.random.Generator.from_seed(42)
```

```python

random_1 = tf.random.Generator.from_seed(42).normal(shape=(3,2))
random_2 = tf.random.Generator.from_seed(42).normal(shape=(3,2))
```

```python
random_1,random_2
```

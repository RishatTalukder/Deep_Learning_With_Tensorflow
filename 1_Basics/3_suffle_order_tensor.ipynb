# Suffling a tensor

Suppose, you have a tensor where the 100 elements are pictures of cats and last 50 elements are pictures of dogs. Your deep learning model is learning to classify cats for the first 100 elements and dogs for the last 50 elements. This can be a problem because the model can learn to classify cats for the first 100 elements and dogs for the last 50 elements. To avoid this, you can shuffle the tensor.

```python
import warnings
import tensorflow as tf

warnings.filterwarnings("ignore")
```

```python
dummy_tenosr = tf.constant([[1, 2], [3, 4], [5, 6]])
dummy_tenosr
```

```python
tf.random.shuffle(dummy_tenosr)
```

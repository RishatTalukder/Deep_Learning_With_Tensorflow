## Tensors

A `tensor` is a `generalization` of vectors and matrices to potentially higher dimensions. Internally, TensorFlow represents tensors as n-dimensional arrays of base datatypes.

there many ways to create `tensors` in `tensorflow` but most common ways are using `tf.constant` and `tf.Variable` functions.

## tf.constant

Here is how you can create a `tensor` using `tf.constant` function


````python
scaler = tf.constant(1)
scaler
````

here, we have created a `tensor` that contains a numpy value `1` and the datatype of the tensor is `int32`.

if you take a closer look you can see that the shape of the tensor is `()` which means it is a `0-D` tensor. So, it's a scalar.

We have a tensor in our hand.

Now you might think why does I named the tensor as `scaler` because it's a `0-D` tensor AKA a `scaler`.

> A `scaler` value is something that has only magnitude and no direction.

To prove it we can see the dimension of the tensor.

````python
scaler.ndim
````

Here you go. You can just use the `variable_name.ndim` to get the dimension of the tensor.

Now, let's see how to create a `1-D` tensor.

````python
vector = tf.constant([1, 2, 3])

vector
````

And we have created a `1-D` tensor. We just pass a `list` of values or a `numpy array` to the `tf.constant` function.

And you can see that the tensor is containing a `numpy array` of values `[1, 2, 3]` and the datatype of the tensor is `int32` and the shape of the tensor is `(3,)` which means it is a `1-D` tensor.

Now we can see the dimansions.

````python
vector.ndim
````

and you see that the tensor is a `1-D` tensor.

> A `vector` is a 1-D array of values.

And we can keep going. So let's try top make a `2-D` tensor and also a `3-D` tensor.

````python
matrix = tf.constant([
    [1, 2],
    [3, 4]
])
````

````python
matrix.ndim
````

it's a 2 by 2 tensor. Simply it's a `2-D` array of values is refered to as a `matrix`.

## tf.Variable

So, if you want to create a mutable tensor you can use the `tf.Variable` function.

This is a more powerful version of the `tf.constant` function.

Useing `tf.Variable` function we can do everything that we can do with `tf.constant` function and many more.

````python
vector = tf.Variable([1,2,3,4,5])
vector
````

````python
vector.dtype
````

````python
vector.shape
````

Now, let's try to assign a value to a index of the vector.

````python
vector[1] = 3
````

Annnnd this doesn't work. In tensorflow, assignment doesn't work like normal assignment. You can't use the assignment operator `=` to assign a value to a index of the vector, instead you have to use the `assign` method.

````python
vector[1].assign(32)
vector
````

Assignment is working now. Now, before we move to more wierd things I want to talk about the output of the `tf.Variable` function.

> `<tf.Variable 'Variable:0' shape=(5,) dtype=int32, numpy=array([ 1, 32,  3,  4,  5], dtype=int32)>`

- The first part is the type of tensor `tf.Variable`.
- `Variable:0` means that the variable is the zeroth variable. Tensorflow assigns unique names to tensor which is made with `tf.Variable` function.
- `shape=(5,)` means that the shape of the tensor is 5.
- `dtype=int32` means that the data type of the tensor is `int32`.
- `numpy=array([ 1, 32,  3,  4,  5], dtype=int32)` means that the values of the tensor are `[ 1, 32,  3,  4,  5]` and the data type of the values is `int32`.

Now if you go to the [official documentations](https://www.tensorflow.org/api_docs/python/tf/Variable) you can see that there are a lot a of arguments that you can pass to the `tf.Variable` function and also a lot of methods and attributes that you can use when you are working with the `tf.Variable` function.

You have access to methods like:

- `assign` to assign a value to a index of the vector.
- `assign_add` to add a value to a index of the vector.
- `assign_sub` to subtract a value from a index of the vector.


## Random Tensors

As I said earlier, tensorflow is a powerful and feature rich version of numpy. It can also help us generate random tensors more easily.

````python
random_1 = tf.random.Generator.from_seed(42)
random_1.normal(shape=(3, 2))
````

To generate random numbers we can use the `tf.random.Generator` class and start the random number generator with the `from_seed` method.

This will return a `Generator` object that we can use to generate random numbers.

> The generator object is like a random number generator with different methods and attributes that we can use to generate random numbers from different distributions like the above I'm using the `normal` method to generate random numbers from a normal distribution with the shape of `(3, 2)`. Meaning this should generate 3 rows and 2 columns of random numbers.

> from_seed method takes a `seed` argument which is an integer.

If you don't know that the seed value does here's a short explanation.

We are generating random numbers but these are not pure random values, these and `pseudo random numbers`. Meaning each random number that this generated is dependent on the `seed` value that we pass to the `from_seed` method.

Seed define the random order of the random numbers that are generated.

So, if we set the seed, the same random number will keep being generated. If we don't then the random order will keep on changing and we will get different random numbers each time we run a file.

It is a best practice to set the seed value to a fixed value so that the random numbers are always the same during production because if the number keeps on changing any bug will be very hard to reproduce.

> Again, if you haven't learned numpy or worked with numpy you should learn numpy now, because almost everything that you are doing with tensorflow is based on numpy.

## Shuffling the data

In the process of training a machine learning model or a neural network the `order of data matters a lot`. A model can easily get overfitted(`learning a pettern so good that learning other patterns becomes tough`) so, shuffling the data before training is a must.

To shuffle the data we can use the `tf.random.shuffle` function.

````python
data = random_1.normal(shape=(5, 3))
data
````

````python
tf.random.shuffle(data, seed=42)
````

There are other methods like `tf.random.stateless_shuffle` that can be used to shuffle the data but I think it is better to use `tf.random.shuffle` because it is more performant. But main gist is we are suffling the data before training.

You guys can use the `tf.random` module to generate different types of random numbers with different distributions and also suffle pre-existing data in different ways.

# New Axis

If you want to add a new axis to a tensor you can use the `tf.newaxis` attribute. This is useful when you want to add a new dimension to a tensor.

````python
tensor = tensor[..., tf.newaxis]
tensor
````

`...` operator is used to say all the other dimensions. It does the same thing as `tensor[:, :, :, tf.newaxis]` but it is more readable.

So, instead of writing `tensor[:, :, :, tf.newaxis]` you can just write `tensor[..., tf.newaxis]`.

Now, by the shape of the tensor you can see that the `new axis` is added to the last dimension and this might be a little confusiong because the tensor looks a little freaky. But what happend is that the `1st dimention` values are now `2nd dimention` values that's why you should see that every value is now inside a array. 

> PS: here, tf.newaxis elevates each dimetion to 1 dimension above.


We can do the exact opposite by using the `tf.squeeze` function.

# Tensor Operations

We can do general tensor operations like addition, subtraction, multiplication, etc and like numpy arrays the values of a tensor is broadcasted automatically.

For example.

````python
tensor + 100
````

As you can see, all the values of the tensor are added by 100. This is called broadcasting. So, when we are doing a operation in a tensor the operation is applied to all the elements of the tensor.

So, we can do operarions like

````python
# print(tensor + 100)
# print(tensor - 100)
# print(tensor * 100)
# print(tensor / 100)
````

Another special operation is the `@` operator. This is called matrix multiplication.

````python
tensor1 = tf.constant(np.random.randint(0, 100, size=(2, 3)))
tensor2 = tf.constant(np.random.randint(0, 100, size=(3, 2)))

tensor1, tensor2
````

Now we can do matrix multiplication.

````python
tensor1 @ tensor2
````

**Rules for matrix multiplication**:

- Inner dimensions must match.

## Some more built in operations and functions

You can do a lot of operations and functions in tensorflow. Here are some of them.

> PS: as every tensor is internally a numpy array you can extract the numpy array from the tensor using the `.numpy()` method.

````python
print("--- Information Functions ---")
# 1. Get shape, data type, and total number of elements
print("Shape of tensor:", tensor.shape)
print("Data type:", tensor.dtype)
print("Total elements:", tf.size(tensor).numpy())
````

````python
print("\n--- Value Aggregation ---")
# 2. Global Minimum and Maximum values
print("Global Min:", tf.reduce_min(tensor).numpy())
print("Global Max:", tf.reduce_max(tensor).numpy())
````

````python
print("\n--- Axis-Specific Aggregation ---")
# 3. Min/Max along columns (axis=0)
print("Min of each column:")
print(tf.reduce_min(tensor, axis=0).numpy())
print("Max of each column:")
print(tf.reduce_max(tensor, axis=0).numpy())
````

````python
# 4. Min/Max along rows (axis=1)
print("Min of each row:")
print(tf.reduce_min(tensor, axis=1).numpy())
print("Max of each row:")
print(tf.reduce_max(tensor, axis=1).numpy())
````

````python
print("\n--- Index Location Functions ---")
# 5. Find the *index position* of the min and max values
# (Flattens the tensor if no axis is specified)
print("Index of global Max:")
print(tf.argmax(tf.reshape(tensor, [-1])).numpy())
print("Index of Max in each column:")
print(tf.argmax(tensor, axis=0).numpy())
````

These are all methods and functions that you can use to aggregate, analyze, and manipulate tensors in tensorflow.
